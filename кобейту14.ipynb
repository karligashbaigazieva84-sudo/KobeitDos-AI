```python
# 🎮 5 есептік көбейту ойыны

import random
import ipywidgets as widgets
from IPython.display import display, HTML

# Ойын мәліметтері
game = {
    "question": 1,
    "score": 0,
    "a": 0,
    "b": 0,
    "correct": 0,
    "mistakes": []
}

# Экрандар
game_screen = widgets.HTML()
game_answer = widgets.Text(
    placeholder="Жауабыңды жаз...",
    description="✏️ Жауап:"
)
game_message = widgets.HTML()


def new_question():

    # 5 есеп біткен болса
    if game["question"] > 5:

        percent = game["score"] * 20

        game_screen.value = f"""
        <div style="
            background:#E8F5E9;
            padding:25px;
            border-radius:18px;
            text-align:center;
        ">
            <h1>🏆 Ойын аяқталды!</h1>
            <h2>Ұпай: {game["score"]} / 5</h2>
            <h2>Нәтиже: {percent}%</h2>
        </div>
        """

        game_answer.layout.display = "none"

        if len(game["mistakes"]) == 0:
            game_message.value = """
            <h2 style="color:green; text-align:center;">
            🌟 Керемет! Барлық есеп дұрыс!
            </h2>
            """
        else:
            mistakes_text = "<br>".join(game["mistakes"])

            game_message.value = f"""
            <div style="
                background:#FFF3E0;
                padding:15px;
                border-radius:15px;
            ">
                <h3>📌 Қиналған есептер:</h3>
                {mistakes_text}
            </div>
            """

        return

    # Жаңа есеп
    game["a"] = random.randint(2, 9)
    game["b"] = random.randint(2, 9)
    game["correct"] = game["a"] * game["b"]

    game_screen.value = f"""
    <div style="
        background:#E3F2FD;
        padding:25px;
        border-radius:18px;
        text-align:center;
    ">
        <h2>🎮 KobeitDos AI ойыны</h2>

        <p style="font-size:18px;">
            Сұрақ {game["question"]} / 5
        </p>

        <h1 style="font-size:40px;">
            {game["a"]} × {game["b"]} = ?
        </h1>

        <p>Жауапты жазып, Enter бас.</p>
    </div>
    """

    game_answer.value = ""
    game_message.value = ""


def check_game(widget):

    # Бос жауап болса
    if game_answer.value.strip() == "":
        return

    try:
        answer = int(game_answer.value)
    except:
        game_message.value = """
        <h3 style="color:#d32f2f;">
        ⚠️ Жауапқа сан жаз.
        </h3>
        """
        return

    # Дұрыс жауап
    if answer == game["correct"]:

        game["score"] += 1

        game_message.value = """
        <h3 style="color:green;">
        ✅ Дұрыс!
        </h3>
        """

    # Қате жауап
    else:

        game["mistakes"].append(
            f'{game["a"]} × {game["b"]} = {game["correct"]}'
        )

        game_message.value = f"""
        <h3 style="color:#d32f2f;">
        ❌ Қате. Дұрыс жауап: {game["correct"]}
        </h3>
        """

    game["question"] += 1

    # Келесі есеп
    new_question()


# Enter арқылы жауап беру
game_answer.on_submit(check_game)

# Экранға шығару
display(game_screen)
display(game_answer)
display(game_message)

# Бірінші есеп
new_question()
```

```python
# 🧠 KobeitDos AI — ДИАГНОСТИКА

import random
import ipywidgets as widgets
from IPython.display import display, HTML

# Диагностика деректері
diag = {
    "questions": [],
    "index": 0,
    "results": {n: {"correct": 0, "total": 0} for n in range(2, 10)}
}

# 2-ден 9-ға дейін әр кестеден 2 есеп
for n in range(2, 10):
    for i in range(2):
        b = random.randint(2, 9)
        diag["questions"].append((n, b))

random.shuffle(diag["questions"])

# Экран
diag_screen = widgets.HTML()

diag_answer = widgets.Text(
    placeholder="Жауабыңды жаз...",
    description="✏️ Жауап:"
)

diag_info = widgets.HTML()


def show_diag_question():

    # Барлық 16 есеп аяқталды
    if diag["index"] >= len(diag["questions"]):
        finish_diagnostic()
        return

    a, b = diag["questions"][diag["index"]]

    diag_screen.value = f"""
    <div style="
        background:#FFF3E0;
        padding:25px;
        border-radius:18px;
        text-align:center;
    ">
        <h2>🧠 KobeitDos AI диагностикасы</h2>

        <p style="font-size:18px;">
            Сұрақ {diag["index"] + 1} / 16
        </p>

        <h1 style="font-size:40px;">
            {a} × {b} = ?
        </h1>

        <p>Жауапты жазып, Enter бас.</p>
    </div>
    """

    diag_answer.value = ""
    diag_info.value = ""


def check_diag(widget):

    if diag_answer.value.strip() == "":
        return

    try:
        answer = int(diag_answer.value)
    except:
        diag_info.value = """
        <h3 style="color:red;">
        ⚠️ Тек сан енгіз.
        </h3>
        """
        return

    a, b = diag["questions"][diag["index"]]
    correct_answer = a * b

    # Осы кестеге бір сұрақ қосылды
    diag["results"][a]["total"] += 1

    # Дұрыс болса
    if answer == correct_answer:
        diag["results"][a]["correct"] += 1

    # Келесі сұрақ
    diag["index"] += 1
    show_diag_question()


def finish_diagnostic():

    global diagnostic_percent
    global weakest_number

    diagnostic_percent = {}

    for n in range(2, 10):

        correct = diag["results"][n]["correct"]
        total = diag["results"][n]["total"]

        if total > 0:
            percent = int((correct / total) * 100)
        else:
            percent = 0

        diagnostic_percent[n] = percent

    # Ең төмен нәтиже
    weakest_number = min(
        diagnostic_percent,
        key=diagnostic_percent.get
    )

    rows = ""

    for n in range(2, 10):

        p = diagnostic_percent[n]

        rows += f"""
        <div style="
            margin:7px;
            padding:10px;
            background:white;
            border-radius:10px;
            font-size:18px;
        ">
            {n}-ге көбейту:
            <b>{p}%</b>
        </div>
        """

    diag_screen.value = f"""
    <div style="
        background:#E8F5E9;
        padding:25px;
        border-radius:18px;
        text-align:center;
    ">

        <h1>📊 Диагностика аяқталды!</h1>

        {rows}

        <hr>

        <h2>
            🤖 KobeitDos AI анықтады:
        </h2>

        <h2 style="color:#d32f2f;">
            Саған {weakest_number}-ге көбейтуді
            көбірек жаттықтыру керек.
        </h2>

    </div>
    """

    diag_answer.layout.display = "none"
    diag_info.value = ""


# Enter арқылы тексеру
diag_answer.on_submit(check_diag)

# Экранға шығару
display(diag_screen)
display(diag_answer)
display(diag_info)

# Бірінші сұрақ
show_diag_question()
```

```python
# 🎯 KobeitDos AI — ЖЕКЕ ЖАТТЫҒУ

import random
import ipywidgets as widgets
from IPython.display import display, HTML

# Жеке жаттығу деректері
practice = {
    "question": 1,
    "score": 0,
    "number": weakest_number,
    "b": 0,
    "correct": 0,
    "mistakes": []
}

# Экран
practice_screen = widgets.HTML()

practice_answer = widgets.Text(
    placeholder="Жауабыңды жаз...",
    description="✏️ Жауап:"
)

practice_info = widgets.HTML()


def show_practice_question():

    # 5 сұрақ аяқталса
    if practice["question"] > 5:
        finish_practice()
        return

    # Әлсіз кестеден есеп
    practice["b"] = random.randint(2, 9)

    practice["correct"] = (
        practice["number"] * practice["b"]
    )

    practice_screen.value = f"""
    <div style="
        background:#F3E5F5;
        padding:25px;
        border-radius:18px;
        text-align:center;
    ">

        <h2>🎯 Жеке жаттығу</h2>

        <p>
            🤖 KobeitDos AI саған
            <b>{practice["number"]}-ге көбейтуді</b>
            жаттықтыруды ұсынады.
        </p>

        <p style="font-size:18px;">
            Сұрақ {practice["question"]} / 5
        </p>

        <h1 style="font-size:40px;">
            {practice["number"]} × {practice["b"]} = ?
        </h1>

        <p>
            Жауабыңды жазып, Enter бас.
        </p>

    </div>
    """

    practice_answer.value = ""
    practice_info.value = ""


def check_practice(widget):

    if practice_answer.value.strip() == "":
        return

    try:
        answer = int(practice_answer.value)

    except:
        practice_info.value = """
        <h3 style="color:red;">
            ⚠️ Тек сан енгіз.
        </h3>
        """
        return

    # Дұрыс жауап
    if answer == practice["correct"]:

        practice["score"] += 1

    # Қате жауап
    else:

        practice["mistakes"].append(
            f'{practice["number"]} × '
            f'{practice["b"]} = '
            f'{practice["correct"]}'
        )

    practice["question"] += 1

    show_practice_question()


def finish_practice():

    global practice_percent

    practice_percent = practice["score"] * 20

    # Диагностикадағы бастапқы нәтиже
    before_percent = diagnostic_percent[
        practice["number"]
    ]

    difference = practice_percent - before_percent

    if difference > 0:
        message = f"""
        🌟 Нәтижең {difference}% жақсарды!
        """
    elif difference == 0:
        message = """
        👍 Нәтижең тұрақты. Тағы жаттығып көр!
        """
    else:
        message = """
        💪 Бұл кестені тағы бір рет қайталап көрейік.
        """

    practice_screen.value = f"""
    <div style="
        background:#E8F5E9;
        padding:25px;
        border-radius:18px;
        text-align:center;
    ">

        <h1>🎉 Жеке жаттығу аяқталды!</h1>

        <h2>
            {practice["number"]}-ге көбейту
        </h2>

        <p style="font-size:20px;">
            🧠 Диагностика:
            <b>{before_percent}%</b>
        </p>

        <p style="font-size:20px;">
            🎯 Жаттығудан кейін:
            <b>{practice_percent}%</b>
        </p>

        <h2>
            {message}
        </h2>

    </div>
    """

    practice_answer.layout.display = "none"
    practice_info.value = ""


# Enter арқылы жауап беру
practice_answer.on_submit(check_practice)

# Экранға шығару
display(practice_screen)
display(practice_answer)
display(practice_info)

# Бірінші есеп
show_practice_question()
```

```python
# ==========================================================
# 🎮 KobeitDos AI — ЖАҢАРТЫЛҒАН ОЙЫН
# ==========================================================

import random

game17 = {
    "q": 1,
    "score": 0,
    "lives": 3,
    "a": 0,
    "b": 0,
    "correct": 0
}


game_html17 = widgets.HTML()

game_input17 = widgets.Text(
    description="✏️ Жауап:",
    placeholder="Жауабыңды жаз...",
    layout=widgets.Layout(width="350px")
)

game_message17 = widgets.HTML()


# ----------------------------------------------------------
# ОЙЫНДЫ БАСТАУ
# ----------------------------------------------------------

def game_start17():

    game17["q"] = 1
    game17["score"] = 0
    game17["lives"] = 3

    game_input17.value = ""
    game_input17.layout.display = ""

    game_message17.value = """
    <div style="
        background:#E3F2FD;
        padding:12px;
        border-radius:12px;
        text-align:center;
    ">
        🚀 Ойын басталды! Сәттілік!
    </div>
    """

    game_question17()


# ----------------------------------------------------------
# ЖАҢА СҰРАҚ
# ----------------------------------------------------------

def game_question17():

    if game17["q"] > 5 or game17["lives"] <= 0:
        game_finish17()
        return

    a = random.randint(2, 9)
    b = random.randint(2, 9)

    game17["a"] = a
    game17["b"] = b
    game17["correct"] = a * b

    hearts = "❤️" * game17["lives"]

    game_html17.value = f"""
    <div style="
        background:linear-gradient(135deg,#E3F2FD,#F3E5F5);
        padding:25px;
        border-radius:20px;
        text-align:center;
    ">

        <h1>🎮 Көбейту жарысы</h1>

        <div style="
            display:flex;
            justify-content:center;
            gap:35px;
            flex-wrap:wrap;
            font-size:18px;
        ">

            <div>
                📝 Сұрақ:
                <b>{game17["q"]} / 5</b>
            </div>

            <div>
                ⭐ Ұпай:
                <b>{game17["score"]}</b>
            </div>

            <div>
                {hearts}
            </div>

        </div>

        <div style="
            background:white;
            margin:20px auto;
            padding:22px;
            border-radius:18px;
            max-width:400px;
        ">

            <p style="
                margin:0;
                color:#666;
            ">
                Неше болады?
            </p>

            <h1 style="
                font-size:42px;
                color:#3949AB;
                margin:10px;
            ">
                {a} × {b} = ?
            </h1>

        </div>

        <p>
            Жауабыңды жазып,
            <b>Enter</b> бас.
        </p>

    </div>
    """


# ----------------------------------------------------------
# ЖАУАПТЫ ТЕКСЕРУ
# ----------------------------------------------------------

def game_check17(widget):

    text = game_input17.value.strip()

    if text == "":
        return

    try:
        answer = int(text)

    except:

        game_message17.value = """
        <div style="
            background:#FFF3E0;
            padding:12px;
            border-radius:12px;
            text-align:center;
        ">
            ⚠️ Жауапты санмен жаз.
        </div>
        """

        game_input17.value = ""
        return


    # ДҰРЫС ЖАУАП
    if answer == game17["correct"]:

        game17["score"] += 1

        game_message17.value = """
        <div style="
            background:#E8F5E9;
            padding:12px;
            border-radius:12px;
            text-align:center;
            font-size:18px;
        ">
            🌟 Дұрыс! Жарайсың!
        </div>
        """


    # ҚАТЕ ЖАУАП
    else:

        game17["lives"] -= 1

        correct = game17["correct"]

        game_message17.value = f"""
        <div style="
            background:#FFEBEE;
            padding:12px;
            border-radius:12px;
            text-align:center;
            font-size:18px;
        ">
            💡 Дұрыс жауап:
            <b>{correct}</b>
        </div>
        """


    game17["q"] += 1

    game_input17.value = ""

    game_question17()


# Enter арқылы тексеру
game_input17.on_submit(game_check17)


# ----------------------------------------------------------
# ОЙЫН СОҢЫ
# ----------------------------------------------------------

def game_finish17():

    score = game17["score"]

    game_input17.layout.display = "none"


    if score == 5:

        level = "🏆 Көбейту шебері!"
        message = "Тамаша! Барлық жауап дұрыс!"

    elif score >= 3:

        level = "🌟 Жарайсың!"
        message = "Жақсы нәтиже! Тағы ойнап көр."

    else:

        level = "💪 Жаттығуды жалғастыр!"
        message = "Көбейту кестесін тағы қайталайық."


    game_html17.value = f"""
    <div style="
        background:linear-gradient(135deg,#FFF8E1,#E8F5E9);
        padding:30px;
        border-radius:20px;
        text-align:center;
    ">

        <div style="font-size:55px;">
            🎉
        </div>

        <h1>
            Ойын аяқталды!
        </h1>

        <h2>
            ⭐ {score} / 5
        </h2>

        <h2 style="color:#3949AB;">
            {level}
        </h2>

        <p style="font-size:18px;">
            {message}
        </p>

    </div>
    """


    game_message17.value = ""


# ----------------------------------------------------------
# ОЙЫН БЛОГЫ
# ----------------------------------------------------------

game_box17 = widgets.VBox([
    game_html17,
    game_input17,
    game_message17
])

game_box17.layout.display = "none"


# ==========================================================
# ПЛАТФОРМАҒА ЖАҢА ОЙЫНДЫ ҚОСУ
# ==========================================================

KobeitDos_AI.children = (
    app_title,
    student_box5,
    app_menu,
    home_box14,
    learn_box2,
    game_box17,
    diag_box2,
    practice_box3,
    result_full15,
    class_result_full13
)


# ==========================================================
# МӘЗІРДІ ЖАҢАРТУ
# ==========================================================

app_menu.unobserve(
    app_change15,
    names="value"
)


def app_change17(change):

    section = change["new"]

    home_box14.layout.display = "none"
    learn_box2.layout.display = "none"
    game_box17.layout.display = "none"
    diag_box2.layout.display = "none"
    practice_box3.layout.display = "none"
    result_full15.layout.display = "none"
    class_result_full13.layout.display = "none"


    if section == "home":

        home_box14.layout.display = ""


    elif section == "learn":

        learn_box2.layout.display = ""


    elif section == "game":

        game_box17.layout.display = ""
        game_start17()


    elif section == "diagnostic":

        diag_box2.layout.display = ""
        diagnostic_start2()


    elif section == "practice":

        practice_box3.layout.display = ""
        practice_start3()


    elif section == "result":

        result_full15.layout.display = ""
        show_result5()


    elif section == "class_result":

        class_result_full13.layout.display = ""
        show_class_result12()
        show_class_chart13()


app_menu.observe(
    app_change17,
    names="value"
)


# Қазіргі бөлімді қайта көрсету
app_change17({
    "new": app_menu.value
})


print("✅ Жаңа ойын бөлімі қосылды.")
```
